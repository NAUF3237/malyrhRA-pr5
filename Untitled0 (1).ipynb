```python
# ============================================================
# Практическая работа: Корреляционный анализ (Titanic)
# ============================================================

# 1. Загрузка библиотек
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Загрузка файла
df = pd.read_csv('titanic.csv')
print("Размер исходного датасета:", df.shape)
print(df.head())

# 3. Выбор признаков и целевой переменной
features = ['Pclass', 'Age', 'Sex', 'Fare', 'SibSp', 'Parch']
target = 'Survived'

data = df[features + [target]].copy()

# 4. Проверка пропущенных значений
print("\n--- Пропуски до обработки ---")
print(data.isnull().sum())

# 5. Обработка пропусков в Age (заполняем медианой)
data['Age'] = data['Age'].fillna(data['Age'].median())

# 6. Преобразование Sex в числовой вид
data['Sex'] = data['Sex'].map({'male': 0, 'female': 1})

# 7. Проверка подготовленной таблицы
print("\n--- Информация о таблице ---")
data.info()
print("\n--- Пропуски после обработки ---")
print(data.isnull().sum())
print("\n--- Первые строки ---")
print(data.head())

# 8. Расчёт корреляционной матрицы
corr_matrix = data.corr().round(2)
print("\n--- Корреляционная матрица (округление до 2 знаков) ---")
print(corr_matrix)

# Анализ корреляции с Survived (без самой себя)
survived_corr = corr_matrix['Survived'].drop('Survived')
print("\n--- Корреляция признаков с Survived ---")
print(survived_corr.sort_values(ascending=False))

# Наиболее сильная положительная корреляция
max_pos = survived_corr.idxmax()
print(f"\nНаиболее сильная положительная корреляция с Survived: {max_pos} ({survived_corr[max_pos]})")

# Наиболее сильная отрицательная корреляция
max_neg = survived_corr.idxmin()
print(f"Наиболее сильная отрицательная корреляция с Survived: {max_neg} ({survived_corr[max_neg]})")

# 9. Построение тепловой карты
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,        # значения коэффициентов внутри ячеек
    cmap='coolwarm',   # цветовая палитра
    fmt='.2f',         # формат вывода — 2 знака после запятой
    linewidths=0.5,
    vmin=-1, vmax=1,
    square=True,
    cbar_kws={'label': 'Коэффициент корреляции'}
)
plt.title('Корреляционная матрица признаков Titanic', fontsize=14)
plt.tight_layout()
plt.show()

# 10. Сравнение с машинным обучением (формирование X и y)
X = data[features]
y = data[target]
print("\n--- Признаки X ---")
print(X.head())
print("\n--- Целевая переменная y ---")
print(y.head())
```
